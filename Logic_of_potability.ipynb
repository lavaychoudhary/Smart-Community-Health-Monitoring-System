```python
import pandas as pd
import numpy as np

# 1. Load the dataset
file_path = "/content/Smart health prediction dataset.xlsx"
df = pd.read_excel(file_path)

# 2. Define the conditions for UNSAFE water (Potability = 0)
# Adjust these thresholds based on the specific strictness required for project
unsafe_conditions = (
    (df['Lead Concentration (µg/L)'] > 10) |   # WHO limit: 10 µg/L
    (df['Bacteria Count (CFU/mL)'] > 500) |    # High bacterial contamination
    (df['Turbidity (NTU)'] > 5) |              # Visible turbidity, WHO: < 5
    (df['pH Level'] < 6.5) |                   # Acidic water
    (df['pH Level'] > 8.5) |                   # Highly alkaline water
    (df['Nitrate Level (mg/L)'] > 50) |        # WHO limit: 50 mg/L
    (df['Dissolved Oxygen (mg/L)'] < 3) |      # Stagnant/highly polluted water
    (df['Contaminant Level (ppm)'] > 10)       # General high contaminant threshold

)

# 3. Apply the labels: 0 for Unsafe (matches conditions), 1 for Safe (Potable)
df['Potability'] = np.where(unsafe_conditions, 0, 1)

# 4. Check the distribution of the new labels
print("Label Distribution (Count):")
print(df['Potability'].value_counts())
print("\nLabel Distribution (%):")
print(df['Potability'].value_counts(normalize=True) * 100)

# 5. Save the prepared dataset for ANN training
output_file = "Smart_health_prediction_dataset_1.csv"
df.to_csv(output_file, index=False)
print(f"\nSuccessfully labeled data and saved to {output_file}")
```
